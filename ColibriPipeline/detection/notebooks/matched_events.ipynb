# Matched-event viewer

Interactive explorer for the detection module's matched events. Point it at a
`.../matched` directory (the one written by `detection.rerun` +
`detection.coincidence`, containing `<ts>-Tier<N>/` subdirs of `det_*.txt`
files). The widget gives a Tier filter, a significance-min slider, and
Prev/Next buttons; each event overlays the participating telescopes'
full-minute light curves (REDBIRD red, GREENBIRD green, BLUEBIRD blue) with the
dip window highlighted.

Requires the optional notebook extras:

```bash
pip install -r requirements-notebook.txt
```

Full-minute curves are pulled from each scope's `<minute_key>_stars.npy` via
`lightcurve_analysis.paths` (set `COLIBRI_ENV=sim` / `COLIBRI_SIM_ROOT`, or pass
`archive_roots={scope: night_dir}`). If a stars file can't be found, the viewer
falls back to the det-window arrays.
